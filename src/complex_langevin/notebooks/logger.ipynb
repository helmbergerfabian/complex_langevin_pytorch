First define a SimState instance. It will store the current state of the simulation. If inherits from a logger class _Logger_,  
which can be used to print logs for debugging. For (no) logging set the environment variable **CL_VERBOSE** to ("0") "1".  
Set the variable **before** you import anything from the complex_langevin package

```python
import os
os.environ["CL_VERBOSE"]= "0"

from complex_langevin.simulation.state import SimState
state = SimState()
```

No logging should require close to zero time, to test this, restart the kernel with **CL_VERBOSE = "0"**. On my machine, i get $\approx$ 100ns, which is totally acceptable.

```python
%%timeit
state.log("ASD")
```
